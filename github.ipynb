```python
from github import Github
```

```python
g = Github("")
```

```python
repo = g.get_repo("sec05/Tech-Track")
print(f"Repository: {repo.full_name}")
```

```python
treningRepos = g.search_repositories(query="stars:>10000", sort="stars", order="desc")
for repo in treningRepos[:10]:
    print(f"repo : {repo.full_name}")
    for commit in repo.get_commits()[:5]:
        print(f"Commit: {commit.sha}, Message: {commit.commit.message}")
        for file in commit.files:
            print(f"File: {file.filename}, Changes: {file.changes}")
```
