```python
#find each a class that contains a URL visible on the search page (located in WlydOe)
aiParse.find_all("a", class_ = "WlydOe", href = True)
```

```python
# add all links into list
aiLinks = []
for link in aiParse.find_all("a", class_ = "WlydOe", href = True):
    aiLinks.append(link['href'])
print(aiLinks)
# scrape for all websites listed in artificial intelligence news page
aiDictionary = {}
for page in aiLinks:
    aiSingleScrape:ScrapeApiResponse = scrapfly.resilient_scrape(scrape_config=ScrapeConfig(url=page), retry_on_status_code=[500])
    aiDictionary.update({page : aiSingleScrape.scrape_result['content']})
print(aiDictionary)
```
